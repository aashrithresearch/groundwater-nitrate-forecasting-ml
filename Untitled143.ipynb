```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from xgboost import XGBRegressor
```

```python
fname = '/content/drive/MyDrive/resultphyschem.csv'

df = pd.read_csv(fname, low_memory=False)

print(df.shape)
```

```python
df = df[df["CharacteristicName"].str.contains(
    "nitrate",
    case=False,
    na=False
)]
```

```python
df = df[[
    "MonitoringLocationIdentifier",
    "ActivityStartDate",
    "ResultMeasureValue"
]]
```

```python
df["ActivityStartDate"] = pd.to_datetime(
    df["ActivityStartDate"],
    errors="coerce"
)

df["ResultMeasureValue"] = pd.to_numeric(
    df["ResultMeasureValue"],
    errors="coerce"
)

df = df.dropna()

df = df[df["ResultMeasureValue"] >= 0]
```

```python
lower = df["ResultMeasureValue"].quantile(0.01)
upper = df["ResultMeasureValue"].quantile(0.99)

df = df[
    df["ResultMeasureValue"].between(lower, upper)
]
```

```python
station_counts = df.groupby(
    "MonitoringLocationIdentifier"
).size()

good_stations = station_counts[
    station_counts >= 24
].index

df = df[
    df["MonitoringLocationIdentifier"].isin(good_stations)
]
```

```python
df = df.sort_values([
    "MonitoringLocationIdentifier",
    "ActivityStartDate"
])
```

```python
station_series = []

for station_id, group in df.groupby(
    "MonitoringLocationIdentifier"
):

    group = group.sort_values("ActivityStartDate")
    group = group.set_index("ActivityStartDate")

    ts = group["ResultMeasureValue"].resample("M").mean()

    if ts.dropna().shape[0] >= 24:
        station_series.append(ts)

print("Usable stations:", len(station_series))
```

```python
X_xgb = []
y_xgb = []

for series in station_series:

    series = series.dropna()

    if len(series) < 12:
        continue

    for i in range(6, len(series)):

        lag_1 = series.iloc[i-1]
        lag_2 = series.iloc[i-2]
        lag_3 = series.iloc[i-3]

        rolling_mean_3 = series.iloc[i-3:i].mean()
        rolling_std_3 = series.iloc[i-3:i].std()

        month = series.index[i].month

        X_xgb.append([
            lag_1,
            lag_2,
            lag_3,
            rolling_mean_3,
            rolling_std_3,
            month
        ])

        y_xgb.append(series.iloc[i])
```

```python
X_xgb = np.array(X_xgb)
y_xgb = np.array(y_xgb)
```

```python
print("X shape:", X_xgb.shape)
print("y shape:", y_xgb.shape)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_xgb,
    y_xgb,
    test_size=0.2,
    random_state=42
)
```

```python
model = XGBRegressor(
    n_estimators=500,
    max_depth=4,
    learning_rate=0.03,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    eval_metric="rmse",
    random_state=42
)
```

```python
model.fit(
    X_train,
    y_train,
    eval_set=[
        (X_train, y_train),
        (X_val, y_val)
    ],
    verbose=10
)
```

```python
train_pred = model.predict(X_train)
val_pred = model.predict(X_val)
```

```python
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

print("XGBOOST RESULTS")

print("\nTrain Metrics")
print("MAE:", mean_absolute_error(y_train, train_pred))
print("RMSE:", rmse(y_train, train_pred))
print("R2:", r2_score(y_train, train_pred))

print("\nValidation Metrics")
print("MAE:", mean_absolute_error(y_val, val_pred))
print("RMSE:", rmse(y_val, val_pred))
print("R2:", r2_score(y_val, val_pred))
```

```python
threshold = 0.5  # mg/L tolerance

train_acc = np.mean(
    np.abs(y_train - train_pred) <= threshold
)

val_acc = np.mean(
    np.abs(y_val - val_pred) <= threshold
)
```

```python
print("Train Accuracy:", train_acc)
print("Validation Accuracy:", val_acc)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    y_val[:100],
    label="Actual"
)

plt.plot(
    val_pred[:100],
    label="Predicted"
)

plt.title("XGBoost: Actual vs Predicted")
plt.xlabel("Sample")
plt.ylabel("Nitrate Concentration")
plt.legend()

plt.show()
```
